## 4. AI 的工作

- 輸入：使用者輸入的一句自然語言。
- 輸出格式：`動作,品項,數量`
- 動作只能是：訂購、取消、查詢、統計、未知。
- 品項只能是：美式咖啡、拿鐵、抹茶歐蕾、未知。
- 數量的規則：沒有說數量時預設為 1 杯；能辨識一杯、兩杯、三杯等數量；超出 1～3 杯時交由 Python 程式拒絕。
- 無法判斷時輸出：`未知,未知,1`，系統回覆目前提供的飲品。
- AI 不可以做的事：不能直接修改資料庫、不能自行新增品項、不能決定庫存是否足夠、不能略過數量限制，也不能直接決定訂單是否成功。

AI 只負責理解使用者的意思，實際訂購、取消、庫存計算及金額計算都由 Python 和 SQLite 處理。

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
#
# ============================================================
# 第 2 格：建立資料庫與初始資料
# ============================================================

# 資料庫檔案名稱
DB_NAME = "coffee_shop.db"

# 建立資料庫連線
conn = sqlite3.connect(DB_NAME)

try:
    cursor = conn.cursor()

    # 1. 刪除舊資料表，避免之前測試的資料影響結果
    cursor.execute("DROP TABLE IF EXISTS orders")
    cursor.execute("DROP TABLE IF EXISTS menu")

    # 2. 建立菜單資料表
    cursor.execute("""
        CREATE TABLE menu (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL UNIQUE,
            price INTEGER NOT NULL,
            stock INTEGER NOT NULL
        )
    """)

    # 3. 建立訂單資料表
    cursor.execute("""
        CREATE TABLE orders (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT NOT NULL,
            menu_id INTEGER NOT NULL,
            quantity INTEGER NOT NULL,
            total INTEGER NOT NULL,
            status TEXT NOT NULL DEFAULT '已訂購',
            created_at TEXT DEFAULT CURRENT_TIMESTAMP,
            FOREIGN KEY (menu_id) REFERENCES menu(id)
        )
    """)

    # 4. 加入老師指定的初始菜單資料
    menu_data = [
        ("美式咖啡", 50, 5),
        ("拿鐵", 65, 4),
        ("抹茶歐蕾", 70, 3)
    ]

    cursor.executemany("""
        INSERT INTO menu (name, price, stock)
        VALUES (?, ?, ?)
    """, menu_data)

    # 5. 儲存資料
    conn.commit()

    print("✅ 資料庫建立成功！")
    print("✅ menu 菜單資料表建立成功！")
    print("✅ orders 訂單資料表建立成功！")
    print("✅ 三種飲品初始資料新增成功！")

finally:
    # 6. 關閉資料庫連線
    conn.close()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 品項, 數量, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
#
# ============================================================
# 第 4 格：AI 解讀函式
# 功能：把中文需求轉成 (動作, 品項, 數量, 原始回覆)
# ============================================================

# 允許的動作與飲品
ALLOWED_ACTIONS = ["訂購", "取消", "查詢", "統計"]
ALLOWED_ITEMS = ["美式咖啡", "拿鐵", "抹茶歐蕾"]

# 建立快取，避免同一句話重複呼叫 AI
ai_cache = {}

def parse_ai_response(raw_text):
    # 預設值
    action = "未知"
    item = "未知"
    quantity = 1

    # AI 回覆格式：動作,品項,數量
    parts = re.split(r"[,，]", raw_text.strip())

    # 1. 檢查動作是否在允許清單
    if len(parts) >= 1:
        action_text = parts[0].strip()
        if action_text in ALLOWED_ACTIONS:
            action = action_text

    # 2. 檢查品項是否在允許清單
    if len(parts) >= 2:
        item_text = parts[1].strip()
        if item_text in ALLOWED_ITEMS:
            item = item_text

    # 3. 使用正規表示式取得數量
    if len(parts) >= 3:
        match = re.search(r"\d+", parts[2])
        if match:
            quantity = int(match.group())

    return action, item, quantity


def ai_parse(user_text):
    # 清除前後空白
    user_text = str(user_text).strip()

    # 空白輸入直接回傳未知
    if not user_text:
        return "未知", "未知", 1, "輸入為空白"

    # 相同句子直接使用之前的結果
    if user_text in ai_cache:
        return ai_cache[user_text]

    # 告訴 AI 應該如何解讀
    system_prompt = """
你是校園咖啡廳的訂單語意分析助手。

你的任務是把使用者的中文需求轉成三個欄位：
動作,品項,數量

【動作】
只能輸出：
訂購、取消、查詢、統計、未知

【品項】
只能輸出：
美式咖啡、拿鐵、抹茶歐蕾、未知

【數量】
1. 沒有提到數量，預設為 1。
2. 一杯、兩杯、三杯，轉成 1、2、3。
3. 五杯轉成 5，不要自行改成 3。
4. 不要自行限制數量，由 Python 程式檢查。

【判斷規則】
1. 想買飲料、預約飲料，動作為訂購。
2. 想取消訂單，動作為取消。
3. 詢問剩餘庫存，動作為查詢。
4. 詢問今日各品項銷售情況，動作為統計。
5. 不存在的飲品，品項為未知。
6. 無法理解時，動作與品項都為未知。
7. 統計時，品項為未知，數量為 1。

【輸出規則】
只能輸出一行。
不能有解釋、標題或 Markdown。
格式必須是：
動作,品項,數量

【範例】
我要一杯美式咖啡
訂購,美式咖啡,1

幫我訂兩杯拿鐵
訂購,拿鐵,2

我要三杯抹茶歐蕾
訂購,抹茶歐蕾,3

我要取消拿鐵
取消,拿鐵,1

美式咖啡剩幾杯
查詢,美式咖啡,1

今天各品項賣得怎麼樣
統計,未知,1

我要五杯拿鐵
訂購,拿鐵,5

我要一杯珍珠奶茶
訂購,未知,1

今天天氣真好
未知,未知,1
"""

    # 最多嘗試 3 次
    for attempt in range(3):
        try:
            # 呼叫老師第 1 格建立的 Groq client
            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {
                        "role": "system",
                        "content": system_prompt
                    },
                    {
                        "role": "user",
                        "content": user_text
                    }
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )

            # 取得 AI 原始回覆
            raw_text = response.choices[0].message.content or ""

            # 使用 Python 驗證 AI 的輸出
            action, item, quantity = parse_ai_response(raw_text)

            # 整理成四個回傳值
            result = (action, item, quantity, raw_text)

            # 儲存到快取
            ai_cache[user_text] = result

            # 每次 AI 呼叫後等待 2 秒
            time.sleep(2)

            return result

        except Exception as e:
            error_message = str(e)

            # 遇到 429 額度限制，等待 30 秒
            if "429" in error_message:
                if attempt < 2:
                    print("⚠️ API 使用量限制，30 秒後重試")
                    time.sleep(30)
                    continue

            # 其他錯誤或重試失敗
            print("❌ AI 呼叫失敗：", type(e).__name__)
            return "未知", "未知", 1, "AI 呼叫失敗"

    return "未知", "未知", 1, "AI 重試失敗"


print("✅ 第 4 格：AI 解讀函式建立成功！")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
#
# ============================================================
# 第 5 格：AI 解讀測試
# 測試 9 句中文，與規格書 7-1 一致
# ============================================================

# 測試資料：(類型, 使用者說的話, 預期動作, 預期品項, 預期數量)
test_cases = [
    ("正常", "我要一杯美式咖啡", "訂購", "美式咖啡", 1),
    ("正常", "幫我訂兩杯拿鐵", "訂購", "拿鐵", 2),
    ("正常", "我要三杯抹茶歐蕾", "訂購", "抹茶歐蕾", 3),
    ("正常", "我要取消拿鐵", "取消", "拿鐵", 1),
    ("正常", "美式咖啡剩幾杯", "查詢", "美式咖啡", 1),
    ("進階", "今天各品項賣得怎麼樣", "統計", "未知", 1),
    ("邊界", "我要五杯拿鐵", "訂購", "拿鐵", 5),
    ("錯誤", "我要一杯珍珠奶茶", "訂購", "未知", 1),
    ("錯誤", "今天天氣真好", "未知", "未知", 1)
]

# 儲存測試結果
test_results = []
passed = 0

print("========== 第 5 格：AI 解讀測試 ==========")

# 依序測試 9 句
for i, (test_type, sentence, exp_action, exp_item, exp_qty) in enumerate(test_cases, 1):

    print(f"\n正在測試第 {i} 句：{sentence}")

    # 呼叫第 4 格的 AI 解讀函式
    action, item, quantity, raw = ai_parse(sentence)

    # 預期答案
    expected = (exp_action, exp_item, exp_qty)

    # 實際答案
    actual = (action, item, quantity)

    # 比較是否正確
    is_pass = (expected == actual)

    if is_pass:
        passed += 1
        status = "PASS"
    else:
        status = "FAIL"

    print("預期：", expected)
    print("實際：", actual)
    print("結果：", status)

    # 儲存成表格資料
    test_results.append({
        "編號": i,
        "類型": test_type,
        "使用者說的話": sentence,
        "預期動作": exp_action,
        "預期品項": exp_item,
        "預期數量": exp_qty,
        "實際動作": action,
        "實際品項": item,
        "實際數量": quantity,
        "測試結果": status,
        "AI原始回覆": raw
    })

# 顯示完整測試結果表格
df_test = pd.DataFrame(test_results)

print("\n========== AI 解讀測試結果表 ==========")
display(df_test)

# 統計通過率
total = len(test_cases)
accuracy = passed / total * 100

print("\n========== 測試總結 ==========")
print(f"總測試數：{total} 句")
print(f"通過數量：{passed} 句")
print(f"失敗數量：{total - passed} 句")
print(f"通過率：{accuracy:.1f}%")

if passed == total:
    print("✅ AI 解讀測試全部通過！")
else:
    print("⚠️ 有測試未通過，請查看上方 FAIL 的項目。")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
#
# ============================================================
# 第 6 格：功能函式
# 訂購、取消、查詢庫存、每日上限、銷售統計
# ============================================================

from datetime import datetime
from zoneinfo import ZoneInfo

# 取得台灣今天的日期
def today_date():
    return datetime.now(
        ZoneInfo("Asia/Taipei")
    ).strftime("%Y-%m-%d")


# 取得台灣現在的時間
def now_time():
    return datetime.now(
        ZoneInfo("Asia/Taipei")
    ).strftime("%Y-%m-%d %H:%M:%S")


# ------------------------------------------------------------
# 功能 1：訂購飲品
# ------------------------------------------------------------
def order_drink(student_name, item, quantity=1):

    student_name = str(student_name or "").strip()

    if not student_name:
        return "❌ 訂購需要填寫學生姓名。"

    if item not in ALLOWED_ITEMS:
        return "❌ 找不到這個品項。目前提供：美式咖啡、拿鐵、抹茶歐蕾。"

    if not isinstance(quantity, int) or quantity < 1 or quantity > 3:
        return "❌ 每次只能訂購 1～3 杯。"

    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()
        cursor.execute("BEGIN IMMEDIATE")

        # 查詢飲品資料
        cursor.execute(
            "SELECT id, price, stock FROM menu WHERE name = ?",
            (item,)
        )
        row = cursor.fetchone()

        if row is None:
            return "❌ 找不到指定的飲品。"

        menu_id, price, stock = row

        # 檢查庫存
        if stock < quantity:
            return f"❌ {item} 庫存不足，目前只剩 {stock} 杯。"

        # 查詢該學生今天已訂購的總杯數
        cursor.execute("""
            SELECT COALESCE(SUM(quantity), 0)
            FROM orders
            WHERE student_name = ?
              AND status = '已訂購'
              AND substr(created_at, 1, 10) = ?
        """, (student_name, today_date()))

        used = cursor.fetchone()[0]

        # 每人每日最多 3 杯
        if used + quantity > 3:
            return (
                f"❌ {student_name} 今天已訂 {used} 杯，"
                "每人每天最多只能訂購 3 杯。"
            )

        # 計算總金額
        total = price * quantity

        # 新增訂單
        cursor.execute("""
            INSERT INTO orders
            (student_name, menu_id, quantity, total, status, created_at)
            VALUES (?, ?, ?, ?, '已訂購', ?)
        """, (
            student_name,
            menu_id,
            quantity,
            total,
            now_time()
        ))

        order_id = cursor.lastrowid

        # 扣除庫存
        cursor.execute("""
            UPDATE menu
            SET stock = stock - ?
            WHERE id = ?
        """, (quantity, menu_id))

        conn.commit()

        return (
            f"✅ 訂購成功！\n"
            f"訂單編號：{order_id}\n"
            f"學生姓名：{student_name}\n"
            f"飲品：{item}\n"
            f"數量：{quantity} 杯\n"
            f"總金額：{total} 元\n"
            f"剩餘庫存：{stock - quantity} 杯"
        )

    except Exception as e:
        conn.rollback()
        return f"❌ 訂購失敗：{type(e).__name__}"

    finally:
        conn.close()


# ------------------------------------------------------------
# 功能 2：取消訂單
# ------------------------------------------------------------
def cancel_order(student_name, item):

    student_name = str(student_name or "").strip()

    if not student_name:
        return "❌ 取消訂單需要填寫學生姓名。"

    if item not in ALLOWED_ITEMS:
        return "❌ 找不到這個品項。"

    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()
        cursor.execute("BEGIN IMMEDIATE")

        # 找出該學生該品項最新一筆有效訂單
        cursor.execute("""
            SELECT o.id, o.menu_id, o.quantity
            FROM orders AS o
            JOIN menu AS m ON o.menu_id = m.id
            WHERE o.student_name = ?
              AND m.name = ?
              AND o.status = '已訂購'
            ORDER BY o.id DESC
            LIMIT 1
        """, (student_name, item))

        row = cursor.fetchone()

        if row is None:
            return (
                f"❌ 找不到 {student_name} 的"
                f"{item}有效訂單，無法取消。"
            )

        order_id, menu_id, quantity = row

        # 修改訂單狀態
        cursor.execute("""
            UPDATE orders
            SET status = '已取消'
            WHERE id = ?
        """, (order_id,))

        # 恢復庫存
        cursor.execute("""
            UPDATE menu
            SET stock = stock + ?
            WHERE id = ?
        """, (quantity, menu_id))

        conn.commit()

        return (
            f"✅ 取消成功！\n"
            f"學生姓名：{student_name}\n"
            f"訂單編號：{order_id}\n"
            f"取消品項：{item}\n"
            f"取消數量：{quantity} 杯\n"
            f"已恢復庫存 {quantity} 杯。"
        )

    except Exception as e:
        conn.rollback()
        return f"❌ 取消失敗：{type(e).__name__}"

    finally:
        conn.close()


# ------------------------------------------------------------
# 功能 3：查詢庫存
# ------------------------------------------------------------
def check_stock(item):

    if item not in ALLOWED_ITEMS:
        return "❌ 找不到這個品項。目前提供：美式咖啡、拿鐵、抹茶歐蕾。"

    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        cursor.execute("""
            SELECT name, price, stock
            FROM menu
            WHERE name = ?
        """, (item,))

        row = cursor.fetchone()

        if row is None:
            return "❌ 找不到指定飲品。"

        name, price, stock = row

        return (
            f"☕ 飲品：{name}\n"
            f"單價：{price} 元\n"
            f"剩餘庫存：{stock} 杯"
        )

    finally:
        conn.close()


# ------------------------------------------------------------
# 功能 4：今日銷售統計
# 使用 LEFT JOIN + GROUP BY
# ------------------------------------------------------------
def sales_report():

    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        cursor.execute("""
            SELECT
                m.name,
                m.price,
                m.stock,
                COALESCE(SUM(o.quantity), 0) AS sold,
                COALESCE(SUM(o.total), 0) AS revenue
            FROM menu AS m
            LEFT JOIN orders AS o
                ON m.id = o.menu_id
                AND o.status = '已訂購'
                AND substr(o.created_at, 1, 10) = ?
            GROUP BY
                m.id, m.name, m.price, m.stock
            ORDER BY m.id
        """, (today_date(),))

        rows = cursor.fetchall()

        lines = ["📊 今日咖啡廳銷售統計"]
        lines.append("=" * 30)

        total_revenue = 0

        for name, price, stock, sold, revenue in rows:

            # 今日初始庫存 = 已售 + 剩餘
            initial_stock = stock + sold

            lines.append(
                f"\n品項：{name}\n"
                f"單價：{price} 元\n"
                f"今日初始庫存：{initial_stock} 杯\n"
                f"已售杯數：{sold} 杯\n"
                f"剩餘杯數：{stock} 杯\n"
                f"營收：{revenue} 元"
            )

            total_revenue += revenue

        lines.append("=" * 30)
        lines.append(f"💰 今日總營收：{total_revenue} 元")

        return "\n".join(lines)

    finally:
        conn.close()


# ------------------------------------------------------------
# 功能 5：統一處理使用者需求
# ------------------------------------------------------------
def handle_request(student_name, user_text):

    user_text = str(user_text or "").strip()

    if not user_text:
        return "❌ 請輸入想訂購、取消或查詢的需求。"

    # 呼叫第 4 格的 AI 解讀函式
    action, item, quantity, raw = ai_parse(user_text)

    # AI 無法辨識動作
    if action == "未知":
        return (
            "❌ 無法理解你的需求。\n"
            "目前提供：美式咖啡、拿鐵、抹茶歐蕾。\n"
            "你可以說：我要兩杯拿鐵。"
        )

    # 訂購
    if action == "訂購":
        return order_drink(student_name, item, quantity)

    # 取消
    elif action == "取消":
        return cancel_order(student_name, item)

    # 查詢庫存
    elif action == "查詢":
        return check_stock(item)

    # 今日銷售統計
    elif action == "統計":
        return sales_report()

    return "❌ 無法處理這個需求。"


print("✅ 第 6 格：功能函式建立成功！")
print("✅ 訂購功能")
print("✅ 取消訂單功能")
print("✅ 查詢庫存功能")
print("✅ 每人每日 3 杯限制")
print("✅ 今日銷售統計")
print("✅ 統一需求處理功能")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
#
# ============================================================
# 第 7 格：主流程與流程測試
# 依照規格書 7-2 執行 9 步測試
# ============================================================

# 1. 設定 9 筆流程測試
# 格式：(姓名, 使用者需求, 預期訊息, 預期庫存, 預期營收)
flow_tests = [
    ("小明", "我要兩杯拿鐵",
     "訂購成功", (5, 2, 3), 130),

    ("小華", "我要一杯美式咖啡",
     "訂購成功", (4, 2, 3), 180),

    ("小明", "我要兩杯美式咖啡",
     "每人每天最多", (4, 2, 3), 180),

    ("小明", "我要取消拿鐵",
     "取消成功", (4, 4, 3), 50),

    ("小華", "美式咖啡剩幾杯",
     "剩餘庫存：4", (4, 4, 3), 50),

    ("小美", "我要三杯抹茶歐蕾",
     "訂購成功", (4, 4, 0), 260),

    ("小華", "我要一杯抹茶歐蕾",
     "庫存不足", (4, 4, 0), 260),

    ("小明", "我要五杯拿鐵",
     "每次只能訂購", (4, 4, 0), 260),

    ("店長", "今天各品項賣得怎麼樣",
     "今日總營收：260", (4, 4, 0), 260)
]


# 2. 查詢目前庫存及有效訂單總營收
def get_database_state():

    conn = sqlite3.connect(DB_NAME)

    try:
        cursor = conn.cursor()

        cursor.execute("""
            SELECT stock
            FROM menu
            ORDER BY id
        """)

        stocks = tuple(row[0] for row in cursor.fetchall())

        cursor.execute("""
            SELECT COALESCE(SUM(total), 0)
            FROM orders
            WHERE status = '已訂購'
              AND substr(created_at, 1, 10) = ?
        """, (today_date(),))

        revenue = cursor.fetchone()[0]

        return stocks, revenue

    finally:
        conn.close()


# 3. 檢查測試前的初始資料
initial_stocks, initial_revenue = get_database_state()

print("========== 第 7 格：流程測試 ==========")
print("測試前庫存：", initial_stocks)
print("測試前營收：", initial_revenue)

if initial_stocks != (5, 4, 3) or initial_revenue != 0:
    raise RuntimeError(
        "測試前資料庫不是初始狀態！"
        "請先重新執行第 2 格重建資料庫，"
        "再執行第 3～6 格，最後執行第 7 格。"
    )


# 4. 依序執行 9 步流程測試
flow_results = []
passed = 0

for i, (name, sentence, expected_text,
        expected_stocks, expected_revenue) in enumerate(flow_tests, 1):

    print("\n" + "=" * 45)
    print(f"第 {i} 步")
    print("姓名：", name)
    print("需求：", sentence)

    # 使用第 6 格的主流程函式
    response = handle_request(name, sentence)

    print("\n系統回覆：")
    print(response)

    # 查詢資料庫實際狀態
    actual_stocks, actual_revenue = get_database_state()

    # 判斷三個條件
    message_ok = expected_text in response
    stock_ok = actual_stocks == expected_stocks
    revenue_ok = actual_revenue == expected_revenue

    is_pass = message_ok and stock_ok and revenue_ok

    if is_pass:
        status = "PASS"
        passed += 1
    else:
        status = "FAIL"

    print("\n預期庫存：", expected_stocks)
    print("實際庫存：", actual_stocks)
    print("預期營收：", expected_revenue)
    print("實際營收：", actual_revenue)
    print("測試結果：", status)

    flow_results.append({
        "步驟": i,
        "姓名": name,
        "使用者需求": sentence,
        "預期訊息": expected_text,
        "預期庫存": str(expected_stocks),
        "實際庫存": str(actual_stocks),
        "預期營收": expected_revenue,
        "實際營收": actual_revenue,
        "測試結果": status
    })


# 5. 顯示流程測試結果表格
print("\n========== 流程測試結果表 ==========")

df_flow = pd.DataFrame(flow_results)
display(df_flow)


# 6. 統計測試結果
total = len(flow_tests)
accuracy = passed / total * 100

print("\n========== 測試總結 ==========")
print(f"總測試數：{total} 步")
print(f"通過數量：{passed} 步")
print(f"失敗數量：{total - passed} 步")
print(f"通過率：{accuracy:.1f}%")

if passed == total:
    print("✅ 9 步流程測試全部通過！")
else:
    print("⚠️ 有測試未通過，請檢查 FAIL 項目。")


# 7. 顯示最後的銷售統計
print("\n========== 最終銷售統計 ==========")
print(sales_report())


# 8. 顯示兩張資料表（老師要求）
print("\n========== 最終資料表 ==========")
show_tables()
````
